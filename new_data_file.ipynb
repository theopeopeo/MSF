```python
import pandas as pd
import os

# Define the file paths
data_folder = "data"
file1 = os.path.join(data_folder, "data.xlsx")
file2 = os.path.join(data_folder, "data-2.xlsx")

# Read the Excel files into DataFrames
try:
    df1 = pd.read_excel(file1)
    print(f"Successfully loaded {file1} into a DataFrame.")
except FileNotFoundError:
    print(f"Error: {file1} not found.")
    df1 = None

try:
    df2 = pd.read_excel(file2)
    print(f"Successfully loaded {file2} into a DataFrame.")
except FileNotFoundError:
    print(f"Error: {file2} not found.")
    df2 = None

# Merge the DataFrames if both are loaded
if df1 is not None and df2 is not None:
    # Find common columns between the two DataFrames
    common_columns = list(set(df1.columns).intersection(set(df2.columns)))

    # Merge the DataFrames on the common columns
    merged_df = pd.concat([df1, df2], ignore_index=True)

    # Drop duplicate rows based on the key columns
    merged_df = merged_df.drop_duplicates(subset=['BudgetCode', 'ProjectName', 'DecisionMoment'])

    # Ensure no duplicate headers in the merged DataFrame
    merged_df = merged_df.loc[:, ~merged_df.columns.duplicated()]

    print("\nMerged DataFrame:")
    print(merged_df.head())

    # Optional: Save the merged DataFrame to a CSV file
    merged_df.to_csv(os.path.join(data_folder, "merged_data.csv"), index=False)
    print(f"Merged DataFrame saved as merged_data.csv in {data_folder}.")
else:
    print("One or both DataFrames could not be loaded. Merging skipped.")
```

```python
# Filter df1 and df2 on BudgetCode == 'YE101'
if df1 is not None:
    df1_filtered = df1[df1['BudgetCode'] == 'YE101']
    print("\nFiltered df1:")
    print(df1_filtered)

if df2 is not None:
    df2_filtered = df2[df2['BudgetCode'] == 'YE101']
    df2_filtered = df2[df2['Actuals/forecast'] == 'Actuals']
    print("\nFiltered df2:")
    print(df2_filtered)
    
    # Group by 'whatLVL1Desc' and sum 'Total CHF' for df1_filtered
if df1 is not None and 'whatLVL1Desc' in df1.columns and 'Total CHF' in df1.columns:
    df1_grouped = df1_filtered.groupby('whatLVL1Desc', as_index=False)['Total CHF'].sum()
    print("\nGrouped df1 by 'whatLVL1Desc' with summed 'Total CHF':")
    print(df1_grouped)

# Group by 'whatLVL1Desc' and sum 'Total CHF' for df2_filtered
if df2 is not None and 'whatLVL1Desc' in df2.columns and 'Total CHF' in df2.columns:
    df2_grouped = df2_filtered.groupby('whatLVL1Desc', as_index=False)['Total CHF'].sum()
    
```

```python
# Group by 'whatLVL1Desc' and 'Year' and sum 'Total CHF' for df1_filtered
if df1 is not None and 'whatLVL1Desc' in df1.columns and 'DecisionMoment' in df1.columns and 'Total CHF' in df1.columns:
    df1_filtered['Year'] = df1_filtered['DecisionMoment'].str[:4]  # Extract the year from 'DecisionMoment'
    df1_grouped_by_desc_and_year = df1_filtered.groupby(['whatLVL1Desc', 'Year'], as_index=False)['Total CHF'].sum()
    print("\nGrouped df1 by 'whatLVL1Desc' and 'Year' with summed 'Total CHF':")
    print(df1_grouped_by_desc_and_year)

# Group by 'whatLVL1Desc' and 'Year' and sum 'Total CHF' for df2_filtered
if df2 is not None and 'whatLVL1Desc' in df2.columns and 'DecisionMoment' in df2.columns and 'Total CHF' in df2.columns:
    df2_filtered['Year'] = df2_filtered['DecisionMoment'].str[:4]  # Extract the year from 'DecisionMoment'
    df2_grouped_by_desc_and_year = df2_filtered.groupby(['whatLVL1Desc', 'Year'], as_index=False)['Total CHF'].sum()
    print("\nGrouped df2 by 'whatLVL1Desc' and 'Year' with summed 'Total CHF':")
    print(df2_grouped_by_desc_and_year)
```

```python
import plotly.express as px

# Plot for df1_grouped_by_desc_and_year
if df1 is not None and 'whatLVL1Desc' in df1.columns and 'Year' in df1_filtered.columns:
    categories_df1 = df1_grouped_by_desc_and_year['whatLVL1Desc'].unique()
    for category in categories_df1:
        category_data = df1_grouped_by_desc_and_year[df1_grouped_by_desc_and_year['whatLVL1Desc'] == category]
        fig = px.line(
            category_data,
            x='Year',
            y='Total CHF',
            title=f"Total CHF Over Years for {category} (df1)",
            labels={'Total CHF': 'Total CHF', 'Year': 'Year'}
        )
        fig.show()

# Plot for df2_grouped_by_desc_and_year
if df2 is not None and 'whatLVL1Desc' in df2.columns and 'Year' in df2_filtered.columns:
    categories_df2 = df2_grouped_by_desc_and_year['whatLVL1Desc'].unique()
    for category in categories_df2:
        category_data = df2_grouped_by_desc_and_year[df2_grouped_by_desc_and_year['whatLVL1Desc'] == category]
        fig = px.line(
            category_data,
            x='Year',
            y='Total CHF',
            title=f"Total CHF Over Years for {category} (df2)",
            labels={'Total CHF': 'Total CHF', 'Year': 'Year'}
        )
        fig.show()
```
